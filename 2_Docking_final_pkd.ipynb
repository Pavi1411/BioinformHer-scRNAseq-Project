```python
# -*- coding: utf-8 -*-
"""2. Docking final copy.ipynb

Automatically generated by Colab.

Original file is located at
    https://colab.research.google.com/drive/1frXGVekM9F9qvC0KLEwVWyaO-8IB2xrQ
"""

# ================================================================
# NOTEBOOK 1 — Molecular Docking of Formononetin Ester Derivatives
# Target: AKT1 (PDB: 3O96) | Tool: PyRx (AutoDock Vina)
# ================================================================
# Upload these files to Colab before running:
#   Formononetin_Derivatives.xlsx
#   Binding_affinity.csv
#   Docking_binding_AKT.xlsx
#   AKT1_prepared.pdb
# ================================================================

!pip install rdkit pandas openpyxl matplotlib seaborn numpy py3Dmol scipy --quiet

from rdkit import Chem
from rdkit.Chem import Draw, AllChem, Descriptors, rdMolDescriptors
from rdkit.Chem.Draw import rdMolDraw2D
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import matplotlib.patches as mpatches
import seaborn as sns
import py3Dmol
from scipy import stats
from IPython.display import display, HTML, SVG
import io, os, warnings
warnings.filterwarnings('ignore')

# Publication style
plt.rcParams.update({
    'font.family':       'DejaVu Serif',
    'axes.spines.top':   False,
    'axes.spines.right': False,
    'axes.linewidth':    0.9,
    'axes.labelsize':    12,
    'xtick.labelsize':   9,
    'ytick.labelsize':   10,
})

NAVY   = '#1A3A5C'
STEEL  = '#2E6DA4'
TEAL   = '#1D9E75'
CORAL  = '#D85A30'
GOLD   = '#E8A000'
LTBLUE = '#D6E8F7'
GREY   = '#888780'

print('✅ All dependencies loaded')

# ── Load SMILES from Excel ───────────────────────────────────────
df_smiles = pd.read_excel('Formononetin_Derivatives.xlsx')
df_smiles.columns = ['Compound_ID', 'Acid_Chloride', 'SMILES', 'Product_Name']
df_smiles = df_smiles.dropna(subset=['SMILES']).reset_index(drop=True)

# Validate all SMILES
invalid = []
for _, row in df_smiles.iterrows():
    mol = Chem.MolFromSmiles(row['SMILES'])
    if mol is None:
        invalid.append(row['Compound_ID'])

print(f'✅ Loaded: {len(df_smiles)} compounds')
print(f'✅ Valid SMILES: {len(df_smiles) - len(invalid)} / {len(df_smiles)}')
if invalid:
    print(f'❌ Invalid: {invalid}')
else:
    print('✅ All SMILES valid — no errors')

df_smiles.head()

from rdkit import Chem
from rdkit.Chem import Draw, AllChem
from rdkit.Chem.Draw import rdMolDraw2D
from rdkit.Chem import rdDepictor
from rdkit.Chem.Draw import IPythonConsole
from PIL import Image as PILImage, ImageDraw as PILImageDraw, ImageFont
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch
import numpy as np
import io

rdDepictor.SetPreferCoordGen(True)

# ================================================================
# PART A — Core structure image (formononetin, C-7 site marked)
# This section was missing from the notebook→script export; it
# builds `core_img`, which PART C below combines with the R-group
# table into Figure1_scheme_final.png.
# ================================================================

FORMONONETIN_CORE_SMILES = 'COc1ccc(-c2coc3cc(O)ccc3c2=O)cc1'
core_mol = Chem.MolFromSmiles(FORMONONETIN_CORE_SMILES)
AllChem.Compute2DCoords(core_mol)

# The only phenolic –OH is the C-7 oxygen — the esterification site
c7_o_idx = None
for atom in core_mol.GetAtoms():
    if atom.GetSymbol() == 'O' and atom.GetTotalNumHs() == 1:
        c7_o_idx = atom.GetIdx()
        break

drawer = rdMolDraw2D.MolDraw2DCairo(900, 700)
opts = drawer.drawOptions()
opts.bondLineWidth = 2
opts.padding = 0.12

highlight_atoms = [c7_o_idx] if c7_o_idx is not None else []
highlight_colors = ({c7_o_idx: (0.85, 0.25, 0.15)}
                     if c7_o_idx is not None else {})

rdMolDraw2D.PrepareAndDrawMolecule(
    drawer, core_mol,
    highlightAtoms=highlight_atoms,
    highlightAtomColors=highlight_colors)
drawer.FinishDrawing()
core_struct_img = PILImage.open(
    io.BytesIO(drawer.GetDrawingText())).convert('RGB')

# Caption strip above the structure, same serif publication style
# as the R-group table
cap_h = 70
core_img = PILImage.new(
    'RGB',
    (core_struct_img.width, core_struct_img.height + cap_h),
    (255, 255, 255))
core_img.paste(core_struct_img, (0, cap_h))

draw_cap = PILImageDraw.Draw(core_img)
try:
    font = ImageFont.truetype(
        '/usr/share/fonts/truetype/dejavu/DejaVuSerif-Bold.ttf', 28)
except Exception:
    font = ImageFont.load_default()
caption = 'Formononetin core — R at C-7 (highlighted, red)'
bbox = draw_cap.textbbox((0, 0), caption, font=font)
tw = bbox[2] - bbox[0]
draw_cap.text(((core_img.width - tw) / 2, 15), caption,
              fill=(20, 20, 20), font=font)

print('✅ Core structure image ready (Part A)')

# ================================================================
# PART B — R-group table (50 derivatives, plain manuscript style)
# ================================================================

r_groups = [
    ( 1, "3,4-Difluorobenzoyl chloride",          "3,4-F₂C₆H₃CO–"),
    ( 2, "Pivaloyl chloride",                      "(CH₃)₃CCO–"),
    ( 3, "Nicotinoyl chloride",                    "Pyridin-3-ylCO–"),
    ( 4, "2-Chloronicotinoyl chloride",            "2-ClPyridin-3-ylCO–"),
    ( 5, "Thiophen-2-carbonyl chloride",           "Thiophen-2-ylCO–"),
    ( 6, "Cyclopropanecarbonyl chloride",          "c-C₃H₅CO–"),
    ( 7, "2,4-Dichlorobenzoyl chloride",           "2,4-Cl₂C₆H₃CO–"),
    ( 8, "4-Trifluoromethylbenzoyl chloride",      "4-CF₃C₆H₄CO–"),
    ( 9, "4-Bromobenzoyl chloride",                "4-BrC₆H₄CO–"),
    (10, "Methoxyacetyl chloride",                 "CH₃OCH₂CO–"),
    (11, "Phenylacetyl chloride",                  "PhCH₂CO–"),
    (12, "Tetrahydrothiophen-3-carbonyl chloride", "THT-3-ylCO–"),
    (13, "Acetyl chloride",                        "CH₃CO–"),
    (14, "2-Furoyl chloride",                      "Furan-2-ylCO–"),
    (15, "2-Pyridylcarbonyl chloride",             "Pyridin-2-ylCO–"),
    (16, "4-Chlorobenzoyl chloride",               "4-ClC₆H₄CO–"),
    (17, "3-Cyanobenzoyl chloride",                "3-CNC₆H₄CO–"),
    (18, "Isobutyryl chloride",                    "(CH₃)₂CHCO–"),
    (19, "3-Nitrobenzoyl chloride",                "3-NO₂C₆H₄CO–"),
    (20, "Indole-3-carbonyl chloride",             "Indol-3-ylCO–"),
    (21, "3-Chloropropionyl chloride",             "3-ClCH₂CH₂CO–"),
    (22, "Ethyl chloroformate",                    "ClCO₂Et–"),
    (23, "1-Naphthoyl chloride",                   "1-NaphthylCO–"),
    (24, "3,5-Dimethylbenzoyl chloride",           "3,5-Me₂C₆H₃CO–"),
    (25, "Chloroacetyl chloride",                  "ClCH₂CO–"),
    (26, "2-Methoxybenzoyl chloride",              "2-MeOC₆H₄CO–"),
    (27, "4-Tert-butylbenzoyl chloride",           "4-tBuC₆H₄CO–"),
    (28, "Cyclopentanecarbonyl chloride",          "c-C₅H₉CO–"),
    (29, "2,4,6-Trimethylbenzoyl chloride",        "2,4,6-Me₃C₆H₂CO–"),
    (30, "4-Ethoxybenzoyl chloride",               "4-EtOC₆H₄CO–"),
    (31, "Hexanoyl chloride",                      "n-C₅H₁₁CO–"),
    (32, "3-Methoxybenzoyl chloride",              "3-MeOC₆H₄CO–"),
    (33, "Salicyloyl chloride",                    "2-HOC₆H₄CO–"),
    (34, "4-Cyanobenzoyl chloride",                "4-CNC₆H₄CO–"),
    (35, "2,6-Difluorobenzoyl chloride",           "2,6-F₂C₆H₃CO–"),
    (36, "Pyridazine-3-carbonyl chloride",         "Pyridazin-3-ylCO–"),
    (37, "4-Methylbenzoyl chloride",               "4-MeC₆H₄CO–"),
    (38, "Oxazole-4-carbonyl chloride",            "Oxazol-4-ylCO–"),
    (39, "Benzimidazole-2-carbonyl chloride",      "Benzimidazol-2-ylCO–"),
    (40, "Cyclohexanecarbonyl chloride",           "c-C₆H₁₁CO–"),
    (41, "4-Hydroxybenzoyl chloride",              "4-HOC₆H₄CO–"),
    (42, "Valeryl chloride",                       "n-C₄H₉CO–"),
    (43, "3,5-Dichlorobenzoyl chloride",           "3,5-Cl₂C₆H₃CO–"),
    (44, "Benzoyl chloride",                       "PhCO–"),
    (45, "3-Methylthiobenzoyl chloride",           "3-MeSC₆H₄CO–"),
    (46, "3,4-Dimethoxybenzoyl chloride",          "3,4-(MeO)₂C₆H₃CO–"),
    (47, "4-Fluorobenzoyl chloride",               "4-FC₆H₄CO–"),
    (48, "4-Hydroxy-3-methoxybenzoyl chloride",    "4-OH-3-MeOC₆H₃CO–"),
    (49, "2-Naphthoyl chloride",                   "2-NaphthylCO–"),
    (50, "4-Nitrobenzoyl chloride",                "4-NO₂C₆H₄CO–"),
]

row_h = 0.0150
fig, ax = plt.subplots(figsize=(8.5, 26))
ax.set_xlim(0, 1)
ax.set_ylim(0, 1)
ax.axis('off')
ax.set_facecolor('white')
fig.patch.set_facecolor('white')

xL, xR  = 0.02, 0.98
x_num   = 0.055
x_name  = 0.110
x_rgrp  = 0.635
y0      = 0.992

# Caption
ax.text(0.5, y0,
        'R = acyl chloride reagents used for C-7 '
        'esterification of formononetin (n = 50)',
        ha='center', va='top', fontsize=9,
        color='black', style='italic',
        fontfamily='DejaVu Serif')

# Column headers
y_ch = y0 - 1.8 * row_h
ax.text(x_num,         y_ch, 'No.',
        ha='center', va='center',
        fontsize=9, fontweight='bold',
        color='black',
        fontfamily='DejaVu Serif')
ax.text(x_name + 0.23, y_ch, 'Acid Chloride',
        ha='center', va='center',
        fontsize=9, fontweight='bold',
        color='black',
        fontfamily='DejaVu Serif')
ax.text(x_rgrp + 0.14, y_ch, 'R',
        ha='center', va='center',
        fontsize=9, fontweight='bold',
        style='italic', color='black',
        fontfamily='DejaVu Serif')

# Header border lines
ax.axhline(y_ch + row_h * 0.62,
           xmin=xL, xmax=xR,
           color='black', lw=1.0)
ax.axhline(y_ch - row_h * 0.58,
           xmin=xL, xmax=xR,
           color='black', lw=0.5)

# Rows
for num, name, rgrp in r_groups:
    y = y_ch - num * row_h * 1.04

    # Alternating light grey rows — plain, no colour
    if num % 2 == 0:
        ax.add_patch(FancyBboxPatch(
            (xL, y - row_h * 0.52),
            xR - xL, row_h * 0.98,
            boxstyle='square,pad=0',
            facecolor='#F2F2F2',
            edgecolor='none', zorder=1))

    ax.text(x_num, y, str(num),
            ha='center', va='center',
            fontsize=8, color='black',
            fontfamily='DejaVu Serif',
            zorder=2)
    ax.text(x_name, y, name,
            ha='left', va='center',
            fontsize=7.8, color='black',
            fontfamily='DejaVu Serif',
            zorder=2)
    ax.text(x_rgrp, y, rgrp,
            ha='left', va='center',
            fontsize=7.6, color='black',
            style='italic',
            fontfamily='DejaVu Serif',
            zorder=2)
    # Row separator
    ax.axhline(y - row_h * 0.54,
               xmin=xL, xmax=xR,
               color='#CCCCCC', lw=0.25,
               zorder=1)

# Bottom border
ax.axhline(
    y_ch - 50 * row_h * 1.04 - row_h * 0.6,
    xmin=xL, xmax=xR,
    color='black', lw=0.8)

plt.tight_layout(pad=0.1)
tbl_buf = io.BytesIO()
fig.savefig(tbl_buf, dpi=300,
            bbox_inches='tight',
            facecolor='white', format='png')
plt.close(fig)
tbl_buf.seek(0)
tbl_img = PILImage.open(tbl_buf).copy()

# ================================================================
# PART C — Combine core structure + table
# ================================================================
pad    = 20
W      = 1400
core_r = core_img.resize(
    (W, int(core_img.height * W / core_img.width)),
    PILImage.LANCZOS)
tbl_r  = tbl_img.resize(
    (W, int(tbl_img.height * W / tbl_img.width)),
    PILImage.LANCZOS)

canvas = PILImage.new(
    'RGB',
    (W + 2*pad,
     core_r.height + tbl_r.height + 2*pad),
    (255, 255, 255))
canvas.paste(core_r, (pad, pad))
canvas.paste(tbl_r,  (pad, pad + core_r.height))

canvas.save('Figure1_scheme_final.png',
            dpi=(300, 300))
display(canvas.resize(
    (860,
     int(860 * (core_r.height +
                tbl_r.height + 2*pad) /
             (W + 2*pad)))))
print('✅ Saved: Figure1_scheme_final.png')

# ── Parse raw PyRx CSV ───────────────────────────────────────────
# PyRx format: ligand_name | affinity | rmsd_lb | rmsd_ub
# Best pose = Mode 1 = lowest affinity for each compound

df_raw = pd.read_csv('Binding_affinity.csv', header=None,
                     names=['Ligand','Affinity','RMSD_LB','RMSD_UB',
                            'Extra1','Extra2'],
                     on_bad_lines='skip',
                     encoding='latin-1')

# Keep only valid rows
df_raw = df_raw[df_raw['Ligand'].apply(lambda x: isinstance(x, str))]
df_raw['Affinity'] = pd.to_numeric(df_raw['Affinity'], errors='coerce')
df_raw = df_raw.dropna(subset=['Affinity'])

# Clean compound name from PyRx filename format
# e.g. 3O96_prepared_1-Naphthoyl_chloride_uff_E=345.77 → 1-Naphthoyl chloride
def clean_name(pyrx_str):
    n = str(pyrx_str)
    for prefix in ['3O96_prepared_', '3O96_prepared']:
        if n.startswith(prefix):
            n = n[len(prefix):]
    if '_uff_E=' in n:
        n = n.split('_uff_E=')[0]
    return n.replace('_', ' ').strip()

df_raw['Compound'] = df_raw['Ligand'].apply(clean_name)

# Extract best pose (minimum affinity) per compound
df_best = (df_raw.groupby('Compound', as_index=False)
           .apply(lambda g: g.nsmallest(1, 'Affinity'))
           .reset_index(drop=True))

# Compute pKd and Kd
RT = 0.5923  # kcal/mol at 298K
df_best['pKd']   = -(df_best['Affinity'] / RT) * np.log10(np.e)
df_best['Kd_nM'] = np.exp(df_best['Affinity'] / RT) * 1e9
df_best['pKd']   = df_best['pKd'].round(2)
df_best['Kd_nM'] = df_best['Kd_nM'].round(2)

df_best = df_best.rename(columns={'Affinity': 'dG_kcal_mol'})
df_best = df_best.sort_values('dG_kcal_mol').reset_index(drop=True)
df_best.index += 1

print(f'✅ Best pose extracted for {len(df_best)} compounds')
print(f'   ΔG range: {df_best["dG_kcal_mol"].max():.1f} to {df_best["dG_kcal_mol"].min():.1f} kcal/mol')
df_best[['Compound','dG_kcal_mol','pKd','Kd_nM']].head(10)

# ── Load Sheet2 of Docking_binding_AKT.xlsx ─────────────────────
# Already has clean names + pKd + Kd

df_docking = pd.read_excel('Docking_binding_AKT.xlsx', sheet_name='Sheet2')
df_docking.columns = ['Compound', 'dG_kcal_mol', 'pKd', 'Kd_nM']
df_docking = df_docking.dropna(subset=['dG_kcal_mol'])
df_docking = df_docking.sort_values('dG_kcal_mol').reset_index(drop=True)
df_docking.index += 1
df_docking.index.name = 'Rank'

# Identify lead
LEAD = '4-Ethoxybenzoyl chloride'
df_docking['Is_Lead'] = df_docking['Compound'].str.contains('Ethoxy|ethoxy',
                                                              regex=True)

print(f'✅ Loaded {len(df_docking)} compounds from cleaned Excel')
print(f'\nTop 10 by binding affinity:')
df_docking[['Compound','dG_kcal_mol','pKd','Kd_nM']].head(10)

# ── Figure: All 50 compounds ranked by ΔG ───────────────────────

fig, ax = plt.subplots(figsize=(20, 8))

cpds   = df_docking['Compound'].tolist()
dg     = df_docking['dG_kcal_mol'].tolist()
is_ld  = df_docking['Is_Lead'].tolist()

colours = []
for c, d, il in zip(cpds, dg, is_ld):
    if il:
        colours.append(GOLD)
    elif d <= -11.5:
        colours.append(NAVY)
    elif d <= -10.5:
        colours.append(STEEL)
    else:
        colours.append('#A8C8E8')

x = np.arange(len(cpds))
ax.bar(x, dg, color=colours, edgecolor='white', linewidth=0.3, width=0.78)

# Top 3 labels
for i in range(3):
    ax.text(i, dg[i] - 0.12, f'#{i+1}',
            ha='center', va='top', fontsize=8,
            color='white', fontweight='bold')

# Lead annotation
li = next((i for i,il in enumerate(is_ld) if il), None)
if li is not None:
    ax.annotate(f'★ Lead\n4-Ethoxybenzoyl',
                xy=(li, dg[li]),
                xytext=(li+2.5, dg[li]-1.2),
                fontsize=9, color=GOLD, fontweight='bold',
                arrowprops=dict(arrowstyle='->', color=GOLD, lw=1.3))

# Formononetin reference
ax.axhline(-8.0, color=CORAL, lw=1.2, linestyle='--', alpha=0.7,
           label='Formononetin parent (approx. −8.0 kcal/mol)')

short = [c.replace(' chloride','').replace('benzoyl','benz.')[:16]
         for c in cpds]
ax.set_xticks(x)
ax.set_xticklabels(short, rotation=55, ha='right', fontsize=7.5)
ax.set_ylabel('Binding Free Energy ΔG (kcal/mol)', fontsize=12)
ax.set_ylim(min(dg) - 0.9, -7.5)
ax.set_title(
    'Molecular Docking Results — 50 Formononetin Ester Derivatives vs AKT1 (PDB: 3O96)\n'
    'PyRx (AutoDock Vina) · Best pose (Mode 1) · Ranked by ΔG',
    fontsize=12, fontweight='bold', color=NAVY, pad=8)

legend_els = [
    mpatches.Patch(color=NAVY,      label='ΔG ≤ −11.5 kcal/mol'),
    mpatches.Patch(color=STEEL,     label='ΔG −10.5 to −11.4'),
    mpatches.Patch(color='#A8C8E8', label='ΔG > −10.5 kcal/mol'),
    mpatches.Patch(color=GOLD,      label='Lead — 4-Ethoxybenzoyl'),
]
ax.legend(handles=legend_els, loc='lower right', fontsize=9, framealpha=0.7)
ax.yaxis.grid(True, alpha=0.25, lw=0.5)
ax.set_axisbelow(True)

plt.tight_layout()
plt.savefig('Figure3_docking_ranked.png',
            dpi=300, bbox_inches='tight', facecolor='white')
print('✅ Saved: Figure3_docking_ranked.png')
plt.show()

# ── Figure: All 50 compounds ranked by ΔG ───────────────────────

fig, ax = plt.subplots(figsize=(20, 8))

cpds   = df_docking['Compound'].tolist()
dg     = df_docking['dG_kcal_mol'].tolist()
is_ld  = df_docking['Is_Lead'].tolist()

colours = []
for c, d, il in zip(cpds, dg, is_ld):
    if d <= -11.5:
        colours.append(NAVY)
    elif d <= -10.5:
        colours.append(STEEL)
    else:
        colours.append('#A8C8E8')

x = np.arange(len(cpds))
ax.bar(x, dg, color=colours, edgecolor='white', linewidth=0.3, width=0.78)

# Formononetin reference line
ax.axhline(-8.0, color=CORAL, lw=1.2, linestyle='--', alpha=0.7,
           label='Formononetin parent (approx. −8.0 kcal/mol)')

short = [c.replace(' chloride','').replace('benzoyl','').strip()[:16]
         for c in cpds]
ax.set_xticks(x)
ax.set_xticklabels(short, rotation=55, ha='right', fontsize=7.5)
ax.set_ylabel('Binding Free Energy ΔG (kcal/mol)', fontsize=12)
ax.set_ylim(min(dg) - 0.9, -7.5)
ax.set_title(
    'Molecular Docking Results — 50 Formononetin Ester Derivatives vs AKT1 (PDB: 3O96)\n'
    'PyRx (AutoDock Vina) · Best pose (Mode 1) · Ranked by ΔG',
    fontsize=12, fontweight='bold', color=NAVY, pad=8)

legend_els = [
    mpatches.Patch(color=NAVY,      label='ΔG ≤ −11.5 kcal/mol'),
    mpatches.Patch(color=STEEL,     label='ΔG −10.5 to −11.4'),
    mpatches.Patch(color='#A8C8E8', label='ΔG > −10.5 kcal/mol'),
]
ax.legend(handles=legend_els, loc='lower right', fontsize=9, framealpha=0.7)
ax.yaxis.grid(True, alpha=0.25, lw=0.5)
ax.set_axisbelow(True)

plt.tight_layout()
plt.savefig('Figure3_docking_ranked.png',
            dpi=300, bbox_inches='tight', facecolor='white')
print('✅ Saved: Figure3_docking_ranked.png')
plt.show()

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# ── Left: scatter ────────────────────────────────────────────────
ax1 = axes[0]
ax1.scatter(df_docking['dG_kcal_mol'], df_docking['pKd'],
            color=STEEL, s=65, alpha=0.72,
            edgecolors='white', linewidth=0.5,
            label='Formononetin ester derivatives')

slope, intercept, r, p, _ = stats.linregress(
    df_docking['dG_kcal_mol'], df_docking['pKd'])
x_line = np.linspace(df_docking['dG_kcal_mol'].min(),
                      df_docking['dG_kcal_mol'].max(), 100)
ax1.plot(x_line, slope*x_line + intercept,
         color=CORAL, lw=1.5, linestyle='--',
         label=f'r = {r:.4f}')

ax1.set_xlabel('ΔG (kcal/mol)', fontsize=11)
ax1.set_ylabel('pKd', fontsize=11)
ax1.set_title('(a)  ΔG vs pKd\nΔG = RT ln Kd, T = 298 K',
              fontsize=11, fontweight='bold', color=NAVY)
ax1.legend(fontsize=8.5, framealpha=0.5)
ax1.text(0.05, 0.95, f'r = {r:.4f}\np < 0.001',
         transform=ax1.transAxes, fontsize=9, va='top',
         bbox=dict(boxstyle='round,pad=0.3', facecolor=LTBLUE,
                   edgecolor=NAVY, alpha=0.85))
ax1.yaxis.grid(True, alpha=0.25, lw=0.5)
ax1.set_axisbelow(True)

# ── Right: Kd log histogram ───────────────────────────────────────
ax2 = axes[1]
log_kd = np.log10(df_docking['Kd_nM'].clip(lower=0.01))
ax2.hist(log_kd, bins=18, color=NAVY, alpha=0.75,
         edgecolor='white', linewidth=0.4)

ax2.axvline(0, color=CORAL, lw=1.2, linestyle=':',
            alpha=0.7, label='1 nM threshold')

ticks = [-1, 0, 1, 2, 3]
ax2.set_xticks(ticks)
ax2.set_xticklabels(['0.1 nM', '1 nM', '10 nM',
                      '100 nM', '1000 nM'])
ax2.set_xlabel('Predicted Kd (log scale)', fontsize=11)
ax2.set_ylabel('Number of compounds', fontsize=11)
ax2.set_title('(b)  Kd Distribution\nLog scale',
              fontsize=11, fontweight='bold', color=NAVY)
ax2.legend(fontsize=9, framealpha=0.5)
ax2.yaxis.grid(True, alpha=0.25, lw=0.5)
ax2.set_axisbelow(True)

plt.tight_layout()
plt.savefig('Figure4_pKd_Kd.png',
            dpi=300, bbox_inches='tight', facecolor='white')
print('✅ Saved: Figure4_pKd_Kd.png')
plt.show()

top25 = df_docking.head(25).copy()
hmap  = top25[['dG_kcal_mol','pKd','Kd_nM']].copy()
hmap.insert(0, '−ΔG', -hmap['dG_kcal_mol'])
hmap  = hmap[['−ΔG', 'pKd', 'Kd_nM']]
hmap_norm = (hmap - hmap.min()) / (hmap.max() - hmap.min())

annot = pd.DataFrame({
    '−ΔG':   hmap['−ΔG'].apply(lambda x: f'{x:.1f}'),
    'pKd':   hmap['pKd'].apply(lambda x: f'{x:.2f}'),
    'Kd_nM': hmap['Kd_nM'].apply(lambda x: f'{x:.1f}'),
})

yticks = [f'#{i}  {c[:22]}'
          for i, c in zip(top25.index,
          [c.replace(' chloride','') for c in top25['Compound']])]

fig, ax = plt.subplots(figsize=(8, 12))
sns.heatmap(hmap_norm, annot=annot, fmt='',
            cmap='Blues', ax=ax,
            yticklabels=yticks,
            xticklabels=['−ΔG (kcal/mol)', 'pKd', 'Kd (nM)'],
            linewidths=0.5, linecolor='white',
            cbar_kws={'label': 'Normalised', 'shrink': 0.4},
            annot_kws={'size': 9.5})

ax.set_xticklabels(ax.get_xticklabels(), rotation=0, fontsize=11)
ax.set_yticklabels(ax.get_yticklabels(), rotation=0, fontsize=8.5)
ax.set_title('Docking Results — Top 25 Compounds vs AKT1 (PDB: 3O96)\n'
             'PyRx · AutoDock Vina · Best pose',
             fontsize=11, fontweight='bold', color=NAVY, pad=10)

plt.tight_layout()
plt.savefig('Figure5_heatmap_top25.png',
            dpi=300, bbox_inches='tight', facecolor='white')
print('✅ Saved: Figure5_heatmap_top25.png')
plt.show()

from google.colab import files
uploaded = files.upload()   # a file picker will pop up — select AKT1_prepared.pdb

import os, glob

renamed = 0
for f in glob.glob('/content/*.pdb'):
    base = os.path.basename(f)
    if base == 'AKT1_prepared.pdb':
        continue
    new_base = base.replace(' ', '_')   # turns every space into an underscore
    if new_base != base:
        new_path = os.path.join('/content', new_base)
        os.rename(f, new_path)
        renamed += 1

print(f'✅ Renamed {renamed} files')
pose_files = sorted(glob.glob('/content/*_best_pose.pdb'))
print(f'✅ Found {len(pose_files)} matching pose files')
print(pose_files[:5])

## Cell 11 — Real 3D Visualisation and Interaction Analysis (All 50 Compounds)

# ── Install dependencies ─────────────────────────────────────────
!pip install prolif mdanalysis numpy pandas matplotlib seaborn --quiet

import MDAnalysis as mda
import prolif as plf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from matplotlib.patches import FancyBboxPatch
import seaborn as sns
import py3Dmol
import glob, os, warnings
warnings.filterwarnings('ignore')

print('✅ All dependencies loaded')
print()

# ================================================================
# STEP 1 — Load protein (receptor)
# ================================================================
from rdkit import Chem

PROTEIN_PDB = 'AKT1_prepared.pdb'

rdkit_prot  = Chem.MolFromPDBFile(PROTEIN_PDB, removeHs=False)
protein_mol = plf.Molecule(rdkit_prot)

print(f'✅ Protein loaded: {PROTEIN_PDB}')
print(f'   Residues: {protein_mol.n_residues}')
print()

# ================================================================
# STEP 2 — Find all ligand PDB files
# Convention: *_best_pose.pdb
# ================================================================
# List all best pose PDB files uploaded to Colab
pose_files = sorted(glob.glob('*_best_pose.pdb'))

if not pose_files:
    print('⚠  No *_best_pose.pdb files found in current directory')
    print()
    print('Upload your 50 best pose PDB files to Colab')
    print('Expected format: CompoundName_best_pose.pdb')
    print()
    print('Quick upload all files at once:')
    print('  from google.colab import files')
    print('  files.upload()  # select all 50 PDB files')
else:
    print(f'✅ Found {len(pose_files)} ligand pose files')
    for f in pose_files[:5]:
        print(f'   {f}')
    if len(pose_files) > 5:
        print(f'   ... and {len(pose_files)-5} more')
    print()

# ================================================================
# STEP 3 — Upload helper (run this if files not yet uploaded)
# ================================================================
def upload_pose_files():
    """Run this function to upload all 50 PDB files at once"""
    from google.colab import files
    uploaded = files.upload()
    print(f'✅ Uploaded {len(uploaded)} files')
    return list(uploaded.keys())

# Uncomment to upload:
# pose_files = upload_pose_files()

# ================================================================
# STEP 4 — Run ProLIF interaction analysis on all compounds
# ================================================================
def get_compound_name(pdb_filename):
    """Extract clean compound name from filename"""
    name = os.path.basename(pdb_filename)
    name = name.replace('_best_pose.pdb', '')
    name = name.replace('_chloride', '')
    name = name.replace('_', ' ').strip()
    return name

def fix_ligand_pdb(pdb_path):
    """
    Fix the ligand PDB for MDAnalysis/ProLIF:
    - Add ATOM records if only HETATM
    - Ensure residue name is LIG not UNK
    - Add END record
    """
    with open(pdb_path) as f:
        lines = f.readlines()

    fixed = []
    for line in lines:
        if line.startswith('HETATM') or line.startswith('ATOM'):
            # Fix residue name UNK → LIG
            line = line[:17] + 'LIG' + line[20:]
            # Fix chain ID to L
            line = line[:21] + 'L' + line[22:]
            fixed.append(line)

    if not fixed[-1].strip().startswith('END'):
        fixed.append('END\n')

    tmp_path = pdb_path.replace('.pdb', '_fixed.pdb')
    with open(tmp_path, 'w') as f:
        f.writelines(fixed)
    return tmp_path

# ── NEW: comprehensive VdW radii covering halogens ────────────────
COMPREHENSIVE_VDW = {
    'H': 1.05, 'C': 1.70, 'N': 1.55, 'O': 1.52,
    'F': 1.47, 'Cl': 1.75, 'Br': 1.85, 'I': 1.98,
    'S': 1.80, 'P': 1.80
}

# ── NEW: robust ligand loader with fallback strategies ─────────────
def load_ligand_robust(fixed_pdb):
    """Try progressively more permissive strategies to load a ligand.
    Returns (Molecule, strategy_used)."""
    # Attempt 1: standard load, full vdw radii
    try:
        u_lig = mda.Universe(fixed_pdb)
        u_lig.atoms.guess_bonds(vdwradii=COMPREHENSIVE_VDW)
        return plf.Molecule.from_mda(u_lig), 'standard'
    except Exception:
        pass
    # Attempt 2: force=True (tolerates missing/odd hydrogens)
    try:
        u_lig = mda.Universe(fixed_pdb)
        u_lig.atoms.guess_bonds(vdwradii=COMPREHENSIVE_VDW)
        return plf.Molecule.from_mda(u_lig, force=True), 'force=True'
    except Exception:
        pass
    # Attempt 3: NoImplicit=False (looser valence/kekulization handling)
    try:
        u_lig = mda.Universe(fixed_pdb)
        u_lig.atoms.guess_bonds(vdwradii=COMPREHENSIVE_VDW)
        return plf.Molecule.from_mda(u_lig, NoImplicit=False), 'NoImplicit=False'
    except Exception as e3:
        raise RuntimeError(f'{e3}')

# Run analysis
all_results   = []
failed        = []
fallback_log  = []   # NEW: audit trail of which compounds needed fallbacks

if pose_files:
    print('Running ProLIF interaction analysis...')
    print('-' * 50)

    for pdb_file in pose_files:
        compound = get_compound_name(pdb_file)
        try:
            # Fix PDB format for ProLIF
            fixed_pdb = fix_ligand_pdb(pdb_file)

            # Load ligand — NEW: robust loader with fallbacks
            lig_mol, strategy = load_ligand_robust(fixed_pdb)
            if strategy != 'standard':
                fallback_log.append({'Compound': compound, 'Strategy': strategy})

            # Run fingerprint
            fp = plf.Fingerprint([
                'HBDonor', 'HBAcceptor',
                'Hydrophobic', 'PiStacking',
                'Anionic', 'Cationic', 'CationPi'
            ])
            fp.run_from_iterable([lig_mol], protein_mol)

            # Get results
            df_fp = fp.to_dataframe()

            # Flatten to list of interactions
            interactions = []
            for col in df_fp.columns:
                if df_fp[col].any():
                    # col format: (residue, interaction_type)
                    residue    = str(col[0]) if isinstance(col, tuple) else str(col)
                    inter_type = str(col[1]) if isinstance(col, tuple) else 'Contact'
                    interactions.append(f'{residue}:{inter_type}')

            all_results.append({
                'Compound':     compound,
                'PDB_file':     pdb_file,
                'N_interactions': len(interactions),
                'Interactions': ' | '.join(interactions),
                'HBonds':       sum(1 for x in interactions if 'HB' in x),
                'Hydrophobic':  sum(1 for x in interactions if 'Hydro' in x),
                'PiStack':      sum(1 for x in interactions if 'Pi' in x),
                'Ionic':        sum(1 for x in interactions if
                                    'Anion' in x or 'Cation' in x),
            })
            flag = f' [fallback: {strategy}]' if strategy != 'standard' else ''
            print(f'  ✅ {compound}: {len(interactions)} interactions{flag}')

            # Clean up temp file
            os.remove(fixed_pdb)

        except Exception as e:
            failed.append(compound)
            print(f'  ⚠  {compound}: {e}')

    print()
    print(f'✅ Complete: {len(all_results)} compounds analysed')
    if failed:
        print(f'⚠  Failed:   {len(failed)} compounds: {failed}')

    # NEW: audit trail summary
    if fallback_log:
        print()
        print('⚠️  AUDIT — compounds that needed a fallback loading strategy:')
        print('    (recommend manual visual check via view_3d_complex() before')
        print('     including these in Table 2 / supplementary materials)')
        for entry in fallback_log:
            print(f"    - {entry['Compound']}: {entry['Strategy']}")

# ================================================================
# STEP 5 — 3D Interactive View (one compound at a time)
# ================================================================
def view_3d_complex(ligand_pdb, protein_pdb=PROTEIN_PDB,
                    compound_name='Compound'):
    """
    Show 3D interactive view of protein-ligand complex in Colab.
    Protein = blue cartoon, Ligand = green sticks,
    Binding pocket residues = orange sticks.
    """
    with open(protein_pdb) as f:
        prot_data = f.read()
    with open(ligand_pdb) as f:
        lig_data = f.read()

    view = py3Dmol.view(width=860, height=580)

    # Add protein
    view.addModel(prot_data, 'pdb')
    view.setStyle({'model': 0},
                  {'cartoon': {'color': 'lightblue', 'opacity': 0.75}})

    # Key binding residues — orange sticks
    for res in [179, 211, 228, 270, 290, 292, 296, 438]:
        view.addStyle(
            {'model': 0, 'chain': 'A', 'resi': res},
            {'stick': {'colorscheme': 'orangeCarbon',
                       'radius': 0.20}})

    # Add ligand — green sticks
    view.addModel(lig_data, 'pdb')
    view.setStyle(
        {'model': 1},
        {'stick': {'colorscheme': 'greenCarbon', 'radius': 0.20}})

    # Surface around binding pocket
    view.addSurface(
        py3Dmol.VDW,
        {'opacity': 0.12, 'color': 'white'},
        {'model': 0, 'chain': 'A',
         'resi': [179, 211, 228, 270, 290, 292, 296, 438]})

    # Zoom to ligand
    view.zoomTo({'model': 1})
    view.setBackgroundColor('white')

    print(f'3D view: {compound_name}')
    print('Controls: left-drag=rotate  scroll=zoom  right-drag=pan')
    view.show()

# ── View one compound (change filename as needed) ─────────────────
# view_3d_complex('Ethyl_chloroformate_chloride_best_pose.pdb',
#                  compound_name='Ethyl chloroformate')

# ================================================================
# STEP 6 — Proper 2D Interaction Diagram using RDKit + ProLIF
# ================================================================
def draw_2d_interaction(ligand_pdb, protein_pdb=PROTEIN_PDB,
                        compound_name='Compound',
                        smiles=None):
    """
    Draw a proper 2D ligand interaction diagram.
    Shows ligand 2D structure with residue labels around it.
    Colour coded by interaction type.
    """
    from rdkit import Chem
    from rdkit.Chem import AllChem, Draw
    from rdkit.Chem.Draw import rdMolDraw2D
    from rdkit.Chem import rdDepictor
    import io
    from PIL import Image as PILImage

    # Get interaction data via ProLIF
    fixed_pdb = fix_ligand_pdb(ligand_pdb)
    u_lig     = mda.Universe(fixed_pdb)
    lig_mol   = plf.Molecule.from_mda(u_lig)
    os.remove(fixed_pdb)

    fp = plf.Fingerprint([
        'HBDonor', 'HBAcceptor',
        'Hydrophobic', 'PiStacking',
        'Anionic', 'Cationic'
    ])
    fp.run_from_iterable([lig_mol], protein_mol)
    df_fp = fp.to_dataframe()

    # Collect interacting residues and types
    interactions = {}
    for col in df_fp.columns:
        if df_fp[col].any():
            if isinstance(col, tuple) and len(col) >= 2:
                residue    = str(col[0])
                inter_type = str(col[1])
                interactions[residue] = inter_type

    print(f'\n{compound_name} — {len(interactions)} interactions:')
    for res, itype in interactions.items():
        print(f'  {res}: {itype}')

    if not interactions:
        print('  No interactions detected')
        return

    # Draw 2D interaction schematic
    int_colours = {
        'HBDonor':    '#1D9E75',
        'HBAcceptor': '#1D9E75',
        'Hydrophobic':'#888780',
        'PiStacking': '#2E6DA4',
        'Anionic':    '#D85A30',
        'Cationic':   '#D85A30',
    }

    n_res  = len(interactions)
    radius = 3.8
    cx, cy = 5.0, 5.0

    fig, ax = plt.subplots(figsize=(10, 10))
    ax.set_xlim(0, 10); ax.set_ylim(0, 10); ax.axis('off')

    # Central ligand box
    ax.add_patch(FancyBboxPatch(
        (cx-1.8, cy-0.9), 3.6, 1.8,
        boxstyle='round,pad=0.12',
        facecolor='#D6E8F7', edgecolor='#1A3A5C',
        linewidth=2.0, zorder=3))
    ax.text(cx, cy+0.3, compound_name,
            ha='center', va='center', fontsize=10,
            fontweight='bold', color='#1A3A5C', zorder=4)
    ax.text(cx, cy-0.3, f'ΔG = {df_docking.loc[df_docking["Compound"].str.contains(compound_name.split()[0], case=False, na=False), "dG_kcal_mol"].values[0] if len(df_docking.loc[df_docking["Compound"].str.contains(compound_name.split()[0], case=False, na=False), "dG_kcal_mol"]) > 0 else "N/A"} kcal/mol',
            ha='center', va='center', fontsize=8,
            color='#2E6DA4', style='italic', zorder=4)

    # Place residues in circle around ligand
    angles = np.linspace(0, 2*np.pi, n_res, endpoint=False)
    res_list = list(interactions.items())

    for idx, (angle, (residue, itype)) in enumerate(
            zip(angles, res_list)):
        rx = cx + radius * np.cos(angle)
        ry = cy + radius * np.sin(angle)

        # Keep within bounds
        rx = np.clip(rx, 0.8, 9.2)
        ry = np.clip(ry, 0.8, 9.2)

        col = int_colours.get(itype, GREY)

        # Residue circle
        ax.add_patch(plt.Circle(
            (rx, ry), 0.52,
            color=col, alpha=0.88, zorder=3))
        ax.text(rx, ry, residue,
                ha='center', va='center',
                fontsize=8.5, fontweight='bold',
                color='white', zorder=4)

        # Connection line to ligand
        dx, dy  = cx - rx, cy - ry
        dist_   = np.sqrt(dx**2+dy**2)
        ux_, uy_= dx/dist_, dy/dist_
        x1_ = rx + ux_*0.55
        y1_ = ry + uy_*0.55
        x2_ = cx - ux_*1.85
        y2_ = cy - uy_*0.92

        ls = '-'  if 'HB' in itype else '--'
        lw = 1.8  if 'HB' in itype else 1.1
        ax.plot([x1_,x2_], [y1_,y2_],
                color=col, lw=lw,
                linestyle=ls, alpha=0.7, zorder=2)

        # Interaction type label
        mx_ = (rx + cx) / 2
        my_ = (ry + cy) / 2
        ax.text(mx_, my_+0.2,
                itype.replace('HBDonor','H-bond D')
                     .replace('HBAcceptor','H-bond A')
                     .replace('PiStacking','π-stack'),
                ha='center', fontsize=6.5,
                color=col, style='italic')

    # Legend
    seen_types = set(interactions.values())
    legend_map = {
        'HBDonor':    ('H-bond donor',    '#1D9E75', '-'),
        'HBAcceptor': ('H-bond acceptor', '#1D9E75', '-'),
        'Hydrophobic':('Hydrophobic',      '#888780', '--'),
        'PiStacking': ('π–π stacking',    '#2E6DA4', '--'),
        'Anionic':    ('Electrostatic',    '#D85A30', '-'),
        'Cationic':   ('Electrostatic',    '#D85A30', '-'),
    }
    y_leg = 1.0
    drawn = set()
    for itype in seen_types:
        if itype in legend_map and itype not in drawn:
            lbl, col, ls = legend_map[itype]
            ax.plot([0.2, 0.7], [y_leg]*2,
                    color=col, lw=2.0, linestyle=ls)
            ax.text(0.85, y_leg, lbl,
                    va='center', fontsize=8.5)
            y_leg += 0.38
            drawn.add(itype)

    ax.set_title(
        f'2D Interaction Diagram — {compound_name}\n'
        f'AKT1 (PDB: 3O96) · ProLIF analysis · Real interactions',
        fontsize=11, fontweight='bold', color='#1A3A5C', pad=10)

    plt.tight_layout()
    out_name = f"interaction_{compound_name.replace(' ','_')}.png"
    plt.savefig(out_name, dpi=300, bbox_inches='tight',
                facecolor='white')
    print(f'✅ Saved: {out_name}')
    plt.show()

# ── Draw interaction diagram for one compound ────────────────────
# draw_2d_interaction(
#     'Ethyl_chloroformate_chloride_best_pose.pdb',
#     compound_name='Ethyl chloroformate')

# ================================================================
# STEP 7 — Interaction summary heatmap (all 50)
# ================================================================
def plot_interaction_heatmap(results_list):
    if not results_list:
        print('No results to plot — run Step 4 first')
        return

    df_res = pd.DataFrame(results_list)

    # Build presence matrix
    all_itype_cols = ['HBonds', 'Hydrophobic', 'PiStack', 'Ionic']
    hmap_data      = df_res.set_index('Compound')[all_itype_cols]

    fig, ax = plt.subplots(figsize=(8, 14))
    sns.heatmap(hmap_data,
                annot=True, fmt='d',
                cmap='Blues', ax=ax,
                linewidths=0.5, linecolor='white',
                cbar_kws={'label':'Count','shrink':0.4})
    ax.set_xticklabels(
        ['H-bonds','Hydrophobic','π-stack','Ionic'],
        rotation=0, fontsize=10)
    ax.set_yticklabels(
        ax.get_yticklabels(), rotation=0, fontsize=8)
    ax.set_title(
        'Interaction Profile — All 50 Compounds vs AKT1\n'
        'ProLIF analysis · Counts per interaction type',
        fontsize=11, fontweight='bold',
        color='#1A3A5C', pad=10)

    plt.tight_layout()
    plt.savefig('Figure8_interactions_all50.png',
                dpi=300, bbox_inches='tight', facecolor='white')
    print('✅ Saved: Figure8_interactions_all50.png')
    plt.show()

# ── Call after Step 4 completes ───────────────────────────────────
# plot_interaction_heatmap(all_results)

# ================================================================
# STEP 8 — Save full interaction table
# ================================================================
def save_interaction_table(results_list):
    if not results_list:
        print('No results — run Step 4 first')
        return
    df_out = pd.DataFrame(results_list)
    df_out.to_csv('Table2_interactions_all50.csv', index=False)
    print('✅ Saved: Table2_interactions_all50.csv')
    print(df_out[['Compound','N_interactions',
                   'HBonds','Hydrophobic',
                   'PiStack','Ionic']].to_string())

# ── Call after Step 4 completes ───────────────────────────────────
# save_interaction_table(all_results)

# ================================================================
# QUICK START — run everything for one compound to test first
# ================================================================
print('='*55)
print('QUICK TEST — run on one compound first')
print('='*55)
print()
print('Step 1: Upload all 50 PDB files')
print('  from google.colab import files')
print('  files.upload()')
print()
print('Step 2: View 3D complex')
print('  view_3d_complex(')
print('      "Ethyl_chloroformate_chloride_best_pose.pdb",')
print('      compound_name="Ethyl chloroformate")')
print()
print('Step 3: Real 2D interaction diagram')
print('  draw_2d_interaction(')
print('      "Ethyl_chloroformate_chloride_best_pose.pdb",')
print('      compound_name="Ethyl chloroformate")')
print()
print('Step 4: Run all 50 — uncomment lines in Step 4 above')
print()
print('Step 5: Heatmap and table — uncomment Step 7 and 8')

!pip install prolif mdanalysis numpy pandas matplotlib seaborn --quiet

import MDAnalysis as mda
import prolif as plf
from rdkit import Chem
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch
import seaborn as sns
import py3Dmol
import glob, os, warnings
warnings.filterwarnings('ignore')

print('✅ All dependencies loaded')

PROTEIN_PDB = 'AKT1_prepared.pdb'

# ── Load protein via RDKit directly ──────────────────────────────
# Remove Hs first to avoid valence issues
prot_rdkit = Chem.MolFromPDBFile(
    PROTEIN_PDB,
    removeHs=True,          # remove all explicit H
    sanitize=False,         # skip sanitization
    proximityBonding=False) # use CONECT records only

if prot_rdkit is None:
    print('RDKit could not parse PDB — trying MDAnalysis fallback...')
    u_protein   = mda.Universe(PROTEIN_PDB)
    sel         = u_protein.select_atoms('protein and not name H*')
    protein_mol = plf.Molecule.from_mda(sel, implicit_hydrogens=True)
else:
    Chem.SanitizeMol(prot_rdkit,
                     Chem.SanitizeFlags.SANITIZE_ALL ^
                     Chem.SanitizeFlags.SANITIZE_PROPERTIES)
    protein_mol = plf.Molecule(prot_rdkit)
    print('✅ Protein loaded via RDKit')

print(f'Protein molecule ready: {protein_mol.GetNumAtoms()} atoms')

!pip install prolif mdanalysis numpy pandas matplotlib seaborn --quiet

import MDAnalysis as mda
import prolif as plf
import pandas as pd, numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch
import seaborn as sns, py3Dmol
import glob, os, warnings
warnings.filterwarnings('ignore')

PROTEIN_PDB = 'AKT1_prepared.pdb'

# ── Load only protein residues, exclude all H atoms ──────────────
u_protein   = mda.Universe(PROTEIN_PDB)

# Select: protein residues only, no hydrogens
protein_sel = u_protein.select_atoms('protein and not name H*')

print(f'Selected {protein_sel.n_atoms} protein heavy atoms')
print(f'Residues: {len(protein_sel.residues)}')

# Convert to ProLIF molecule
protein_mol = plf.Molecule.from_mda(
    protein_sel,
    implicit_hydrogens=True)

print('✅ Protein loaded successfully')

# ================================================================
# STEP 2 — Fix and load ligand PDB
# ================================================================
def fix_ligand_pdb(pdb_path):
    """
    Fix ligand PDB for ProLIF compatibility:
    - Change residue name UNK → LIG
    - Ensure chain ID is L
    - Remove hydrogens (avoid valence issues)
    - Add END record
    """
    with open(pdb_path) as f:
        lines = f.readlines()

    fixed = []
    for line in lines:
        if line.startswith('HETATM') or line.startswith('ATOM'):
            atom_name = line[12:16].strip()
            # Skip hydrogens
            if atom_name.startswith('H') or \
               atom_name.startswith('1H') or \
               atom_name.startswith('2H') or \
               atom_name.startswith('3H'):
                continue
            # Fix residue name → LIG
            line = line[:17] + 'LIG' + line[20:]
            # Fix chain ID → L
            line = line[:21] + 'L' + line[22:]
            fixed.append(line)

    if not fixed or not fixed[-1].strip().startswith('END'):
        fixed.append('END\n')

    tmp = pdb_path.replace('.pdb', '_fixed.pdb')
    with open(tmp, 'w') as f:
        f.writelines(fixed)
    return tmp


def load_ligand(pdb_path):
    """Load ligand as ProLIF Molecule — tries multiple methods"""
    fixed_pdb = fix_ligand_pdb(pdb_path)

    # Method 1: RDKit direct
    from rdkit import Chem
    mol = Chem.MolFromPDBFile(
        fixed_pdb, removeHs=True,
        sanitize=False, proximityBonding=True)

    if mol is not None:
        try:
            Chem.SanitizeMol(mol,
                Chem.SanitizeFlags.SANITIZE_ALL ^
                Chem.SanitizeFlags.SANITIZE_PROPERTIES)
            lig_mol = plf.Molecule(mol)
            os.remove(fixed_pdb)
            return lig_mol
        except Exception as e:
            print(f'    RDKit sanitize warning: {e}')

    # Method 2: MDAnalysis fallback
    u_lig   = mda.Universe(fixed_pdb)
    lig_sel = u_lig.select_atoms('all')
    lig_mol = plf.Molecule.from_mda(
        lig_sel, implicit_hydrogens=True)
    os.remove(fixed_pdb)
    return lig_mol


# ── Test on one compound first ───────────────────────────────────
TEST_PDB = 'Ethyl_chloroformate_chloride_best_pose.pdb'

print('Testing ligand loading...')
test_mol = load_ligand(TEST_PDB)
print(f'✅ Ligand loaded: {test_mol.GetNumAtoms()} atoms')

# ================================================================
# STEP 3 — Run ProLIF fingerprint on one compound (test first)
# ================================================================
fp = plf.Fingerprint([
    'HBDonor', 'HBAcceptor',
    'Hydrophobic', 'PiStacking',
    'Anionic', 'Cationic'
])

fp.run_from_iterable([test_mol], protein_mol)
df_test = fp.to_dataframe()

print('✅ ProLIF analysis complete')
print(f'   Interactions detected: {df_test.shape[1]}')
print()

# Show detected interactions
print('Interacting residues:')
for col in df_test.columns:
    if df_test[col].any():
        if isinstance(col, tuple):
            print(f'  {col[0]} — {col[1]}')
        else:
            print(f'  {col}')

# ================================================================
# STEP 4 — 3D interactive view: protein + ligand
# ================================================================
def view_3d_complex(ligand_pdb,
                    protein_pdb=PROTEIN_PDB,
                    compound_name='Compound'):

    with open(protein_pdb) as f:
        prot = f.read()
    with open(ligand_pdb) as f:
        lig  = f.read()

    view = py3Dmol.view(width=860, height=580)

    # Protein — blue cartoon
    view.addModel(prot, 'pdb')
    view.setStyle({'model': 0},
                  {'cartoon': {
                      'color': 'lightblue',
                      'opacity': 0.75}})

    # Key AKT1 binding residues — orange sticks
    for res in [179, 211, 228, 270, 290, 292, 296, 438]:
        view.addStyle(
            {'model': 0, 'chain': 'A', 'resi': res},
            {'stick': {
                'colorscheme': 'orangeCarbon',
                'radius': 0.22}})
        view.addLabel(
            str(res),
            {'fontColor': 'black', 'fontSize': 10,
             'backgroundColor': 'white',
             'backgroundOpacity': 0.6},
            {'model': 0, 'chain': 'A',
             'resi': res, 'atom': 'CA'})

    # Ligand — green sticks
    view.addModel(lig, 'pdb')
    view.setStyle(
        {'model': 1},
        {'stick': {
            'colorscheme': 'greenCarbon',
            'radius': 0.22}})

    # Transparent surface on binding pocket
    view.addSurface(
        py3Dmol.VDW,
        {'opacity': 0.10, 'color': 'grey'},
        {'model': 0, 'chain': 'A',
         'resi': [179, 211, 228, 270,
                  290, 292, 296, 438]})

    view.zoomTo({'model': 1})
    view.setBackgroundColor('white')

    print(f'3D view: {compound_name}')
    print('Rotate=left-drag  Zoom=scroll  Pan=right-drag')
    view.show()

# ── Run 3D view ───────────────────────────────────────────────────
view_3d_complex(
    TEST_PDB,
    compound_name='Ethyl chloroformate')

def load_ligand(pdb_path):
    """
    Load ligand as ProLIF Molecule.
    Adds hydrogens via RDKit from SMILES for better interaction detection.
    """
    fixed_pdb = fix_ligand_pdb(pdb_path)

    from rdkit import Chem
    from rdkit.Chem import AllChem

    # Load from PDB
    mol = Chem.MolFromPDBFile(
        fixed_pdb,
        removeHs=True,
        sanitize=False,
        proximityBonding=True)

    os.remove(fixed_pdb)

    if mol is None:
        raise ValueError(f'Could not load {pdb_path}')

    # Sanitize
    try:
        Chem.SanitizeMol(mol,
            Chem.SanitizeFlags.SANITIZE_ALL ^
            Chem.SanitizeFlags.SANITIZE_PROPERTIES)
    except Exception:
        pass

    # Add hydrogens — critical for H-bond detection
    mol = Chem.AddHs(mol, addCoords=True)

    # If no 3D coords on Hs, embed them
    try:
        AllChem.EmbedMolecule(mol,
            AllChem.ETKDGv3())
    except Exception:
        pass

    return plf.Molecule(mol)

# ── Distance-based interaction detection (no ProLIF) ─────────────
# Detects interactions by measuring actual distances
# H-bond: donor-acceptor distance < 3.5 Å
# Hydrophobic: C-C distance < 4.5 Å
# This works directly with your PDB files

import MDAnalysis as mda
from MDAnalysis.analysis import distances
import numpy as np

def detect_interactions_distance(ligand_pdb,
                                  protein_pdb=PROTEIN_PDB,
                                  hbond_cutoff=3.5,
                                  hydrophobic_cutoff=4.5,
                                  pistack_cutoff=5.5):
    """
    Detect protein-ligand interactions purely by distance.
    No ProLIF needed — works with any PDB format.
    """
    # Load both
    u = mda.Universe(protein_pdb)

    # Load ligand coords directly
    with open(ligand_pdb) as f:
        lines = f.readlines()
    lig_atoms = []
    for line in lines:
        if line.startswith('HETATM') or line.startswith('ATOM'):
            atom_name = line[12:16].strip()
            element   = line[76:78].strip() if len(line) > 76 else atom_name[0]
            try:
                x = float(line[30:38])
                y = float(line[38:46])
                z = float(line[46:54])
                lig_atoms.append({
                    'name':    atom_name,
                    'element': element if element else atom_name[0],
                    'pos':     np.array([x,y,z])
                })
            except ValueError:
                continue

    if not lig_atoms:
        print('No ligand atoms found')
        return {}

    lig_positions = np.array([a['pos'] for a in lig_atoms])
    lig_elements  = [a['element'][0].upper() for a in lig_atoms]

    # Protein atoms with positions
    protein_atoms = u.select_atoms('protein')
    prot_pos      = protein_atoms.positions

    # ── Find nearby residues ──────────────────────────────────────
    interactions = {}

    for i, res in enumerate(u.residues):
        if res.resname in ['HOH','WAT','SOL']:
            continue

        res_atoms = res.atoms
        res_pos   = res_atoms.positions
        res_names = [a.name for a in res_atoms]
        res_elem  = [a.element if hasattr(a,'element')
                     and a.element else a.name[0].upper()
                     for a in res_atoms]

        # Min distance between any ligand atom and residue atom
        min_dist = np.inf
        best_lig_idx  = 0
        best_prot_idx = 0

        for li, lpos in enumerate(lig_positions):
            dists = np.linalg.norm(res_pos - lpos, axis=1)
            mi    = np.argmin(dists)
            if dists[mi] < min_dist:
                min_dist      = dists[mi]
                best_lig_idx  = li
                best_prot_idx = mi

        res_key = f'{res.resname} {res.segid} {res.resid}'

        # Classify interaction
        if min_dist > hydrophobic_cutoff:
            continue

        le = lig_elements[best_lig_idx]
        pe = res_elem[best_prot_idx] if best_prot_idx < len(res_elem) else 'C'

        # H-bond: N or O on both sides within 3.5 Å
        if min_dist <= hbond_cutoff:
            # Check if donor-acceptor pair
            hbond_elements = {'N','O','F'}
            lig_hb = any(lig_elements[li] in hbond_elements
                         for li in range(len(lig_atoms)))
            prot_hb= any(e in hbond_elements
                         for e in res_elem)
            if lig_hb and prot_hb:
                interactions[res_key] = ('HBond', round(min_dist,2))
                continue

        # π-stacking: aromatic residues within 5.5 Å
        aromatic_res = ['PHE','TYR','TRP','HIS']
        if res.resname in aromatic_res and min_dist <= pistack_cutoff:
            interactions[res_key] = ('PiStack', round(min_dist,2))
            continue

        # Hydrophobic: C-C contact
        if le == 'C' and pe == 'C' and min_dist <= hydrophobic_cutoff:
            interactions[res_key] = ('Hydrophobic', round(min_dist,2))

    return interactions


# ── Test ──────────────────────────────────────────────────────────
print('Testing distance-based detection...')
test_interactions = detect_interactions_distance(TEST_PDB)

print(f'\nFound {len(test_interactions)} interactions:')
for res, (itype, dist) in test_interactions.items():
    print(f'  {res}: {itype}  ({dist} Å)')

# ── Install PLIP + OpenBabel (system + python bindings) ──────────
!apt-get install -y openbabel python3-openbabel --quiet > /dev/null 2>&1
!pip install plip --quiet

from plip.structure.preparation import PDBComplex

def run_plip(complex_pdb, out_dir):
    """Run PLIP directly via its Python API (no subprocess needed)."""
    os.makedirs(out_dir, exist_ok=True)

    mol = PDBComplex()
    mol.load_pdb(complex_pdb)
    mol.analyze()

    return mol   # return the analyzed complex object for downstream parsing

# ================================================================
# STEP 5 — Proper interaction diagram using PLIP
# PLIP is the gold standard tool used in published papers
# Produces exactly the style in your reference image
# ================================================================

# ── Install PLIP + OpenBabel ──────────────────────────────────────
!apt-get install -y openbabel python3-openbabel --quiet > /dev/null 2>&1
!pip install plip --quiet

import subprocess, os, shutil
from pathlib import Path
from rdkit import Chem
from rdkit.Chem.Draw import rdMolDraw2D
from rdkit.Chem import rdDepictor
from matplotlib.patches import Ellipse, FancyArrowPatch
from matplotlib.lines import Line2D
import matplotlib.patches as mpatches
import matplotlib.pyplot as plt
import numpy as np, math, io
from PIL import Image as PILImage
from plip.structure.preparation import PDBComplex

rdDepictor.SetPreferCoordGen(True)


# ── Step 1: Build complex PDB (protein + ligand together) ─────────
def build_complex_pdb(protein_pdb, ligand_pdb, out_path='complex.pdb'):
    """
    Merge protein and ligand into one PDB file.
    PLIP requires both in a single file.
    Ligand residue renamed to LIG, chain L.
    """
    lines = []

    # Protein ATOM records
    with open(protein_pdb) as f:
        for line in f:
            if line.startswith('ATOM'):
                lines.append(line.rstrip())

    # Ligand HETATM records
    with open(ligand_pdb) as f:
        for line in f:
            if line.startswith('HETATM') or line.startswith('ATOM'):
                atom_name = line[12:16].strip()
                # Skip hydrogens
                if atom_name.startswith('H'):
                    continue
                # Force HETATM, residue=LIG, chain=L, resid=1
                new_line = (
                    'HETATM' +
                    line[6:17] +
                    'LIG' +
                    ' L' +
                    '   1' +
                    line[26:54] +
                    '  1.00  0.00' +
                    '          LIG\n')
                lines.append(new_line.rstrip())

    lines.append('END')

    with open(out_path, 'w') as f:
        f.write('\n'.join(lines) + '\n')

    print(f'✅ Complex PDB written: {out_path}  ({len(lines)} lines)')
    return out_path


# ── Step 2: Run PLIP via Python API (no subprocess/plipcmd needed) ─
def run_plip(complex_pdb, out_dir=None):
    """Run PLIP directly through its Python API and return the
    analyzed PDBComplex object (skips the txt/xml file step)."""
    mol = PDBComplex()
    mol.load_pdb(complex_pdb)
    mol.analyze()
    print(f'✅ PLIP analysis complete — binding sites found: '
          f'{list(mol.interaction_sets.keys())}')
    return mol


# ── Step 3: Extract interactions directly from PDBComplex object ──
def extract_plip_interactions(mol, ligand_resname='LIG'):
    """Pull interactions straight from PLIP's analyzed object —
    no file parsing needed."""
    interactions = {}

    # Find the binding site key matching our ligand
    site_key = None
    for key in mol.interaction_sets:
        if key.startswith(ligand_resname):
            site_key = key
            break
    if site_key is None:
        print('⚠  No binding site found for ligand LIG')
        return interactions

    site = mol.interaction_sets[site_key]

    def add(items, itype, dist_attr='distance'):
        for it in items:
            try:
                resname = it.restype
                resid   = it.resnr
                chain   = it.reschain
                dist    = getattr(it, dist_attr, None)
                res_key = f'{resname} {chain} {resid}'
                if res_key not in interactions:
                    interactions[res_key] = (itype, dist)
            except Exception:
                continue

    add(site.hbonds_pdon,           'HBond',       'distance_ah')
    add(site.hbonds_ldon,           'HBond',       'distance_ah')
    add(site.hydrophobic_contacts,  'Hydrophobic', 'distance')
    add(site.pistacking,            'PiStack',     'distance')
    add(site.pication_laro,         'PiCation',    'distance')
    add(site.pication_paro,         'PiCation',    'distance')
    add(site.saltbridge_lneg,       'SaltBridge',  'distance')
    add(site.saltbridge_pneg,       'SaltBridge',  'distance')
    add(site.halogen_bonds,         'HalogenBond', 'distance')
    add(site.metal_complexes,       'MetalComplex','distance')

    return interactions


# ── Step 4: Full pipeline function ────────────────────────────────
def ligplot_plip(protein_pdb, ligand_pdb, ligand_smiles,
                 compound_name='Compound', dg_val=None):
    """
    Complete LigPlot-style diagram using PLIP interactions.
    Falls back to distance-based if PLIP fails.
    """
    print(f'\nProcessing: {compound_name}')

    # Build complex
    complex_pdb = f'complex_{compound_name.replace(" ","_")}.pdb'
    build_complex_pdb(protein_pdb, ligand_pdb, complex_pdb)

    # Run PLIP via API
    plip_mol = run_plip(complex_pdb)

    # Extract interactions directly from the analyzed object
    interactions = extract_plip_interactions(plip_mol)
    if not interactions:
        print('PLIP gave no results — using distance-based fallback')
        interactions = detect_interactions_distance(ligand_pdb, protein_pdb)
        # Convert to PLIP format
        interactions = {
            k: (v[0], v[1])
            for k, v in interactions.items()}

    print(f'Interactions found: {len(interactions)}')
    for r,(t,d) in interactions.items():
        dist_str = f'{d:.2f} Å' if d else ''
        print(f'  {r}: {t} {dist_str}')

    if not interactions:
        print('No interactions — skipping diagram')
        return

    # Draw ligand structure
    mol = Chem.MolFromSmiles(ligand_smiles)
    mol = Chem.RemoveHs(mol)
    rdDepictor.Compute2DCoords(mol)

    C      = 1400
    drawer = rdMolDraw2D.MolDraw2DCairo(C, C)
    opts   = drawer.drawOptions()
    opts.padding          = 0.20
    opts.bondLineWidth    = 2.5
    opts.backgroundColour = (1, 1, 1)
    drawer.DrawMolecule(mol)
    drawer.FinishDrawing()
    struct_img = PILImage.open(
        io.BytesIO(drawer.GetDrawingText())).copy()

    # Atom pixel positions
    conf    = mol.GetConformer()
    atom_xy = np.array([
        [conf.GetAtomPosition(i).x,
         conf.GetAtomPosition(i).y]
        for i in range(mol.GetNumAtoms())])

    pad  = 0.20
    xmin = atom_xy[:,0].min(); xmax = atom_xy[:,0].max()
    ymin = atom_xy[:,1].min(); ymax = atom_xy[:,1].max()
    mw   = max(xmax-xmin, 1e-6)
    mh   = max(ymax-ymin, 1e-6)
    use  = C*(1-2*pad)
    sf   = use/max(mw,mh)
    xoff = C*pad+(use-mw*sf)/2
    yoff = C*pad+(use-mh*sf)/2

    def m2p(x,y):
        return xoff+(x-xmin)*sf, C-(yoff+(y-ymin)*sf)

    apx   = np.array([m2p(x,y) for x,y in atom_xy])
    heavy = [i for i,a in enumerate(mol.GetAtoms())
             if a.GetAtomicNum()!=1]
    hpx   = apx[heavy]
    cx_   = hpx[:,0].mean()
    cy_   = hpx[:,1].mean()
    rr_   = max(hpx[:,0].max()-hpx[:,0].min(),
                hpx[:,1].max()-hpx[:,1].min())/2
    pr_   = rr_ + C*0.17

    # Colours — matching your reference image exactly
    col_map = {
        'HBond':       '#3CB371',   # green
        'Hydrophobic': '#DA70D6',   # orchid/pink
        'PiStack':     '#FF8C00',   # orange
        'PiCation':    '#FF8C00',   # orange
        'SaltBridge':  '#DC143C',   # crimson
        'WaterBridge': '#4169E1',   # royal blue
        'HalogenBond': '#20B2AA',   # teal
        'MetalComplex':'#DAA520',   # goldenrod
    }
    ls_map = {
        'HBond':       (0,(5,3)),
        'Hydrophobic': (0,(2,3)),
        'PiStack':     (0,(8,3)),
        'PiCation':    (0,(8,3)),
        'SaltBridge':  '-',
        'WaterBridge': (0,(3,1,1,1)),
        'HalogenBond': (0,(6,2)),
        'MetalComplex':'-',
    }
    lbl_map = {
        'HBond':       'Hydrogen bond',
        'Hydrophobic': 'Hydrophobic contact',
        'PiStack':     'π–π stacking',
        'PiCation':    'π–cation',
        'SaltBridge':  'Salt bridge',
        'WaterBridge': 'Water bridge',
        'HalogenBond': 'Halogen bond',
        'MetalComplex':'Metal complex',
    }

    fig, ax = plt.subplots(figsize=(13, 13))
    ax.set_xlim(0,C); ax.set_ylim(0,C)
    ax.set_aspect('equal'); ax.axis('off')
    ax.imshow(struct_img, extent=[0,C,0,C],
              zorder=1, origin='upper')

    n    = len(interactions)
    angs = np.linspace(0,2*np.pi,n,endpoint=False) - np.pi/2
    ow   = C*0.10; oh = C*0.062

    for ang, (rkey,(itype,dist)) in zip(angs,
            list(interactions.items())):
        rx = cx_ + pr_*math.cos(ang)
        ry = cy_ - pr_*math.sin(ang)
        rx = float(np.clip(rx, ow+30, C-ow-30))
        ry = float(np.clip(ry, oh+30, C-oh-30))

        col = col_map.get(itype,'#778899')
        ls  = ls_map.get(itype,(0,(4,3)))
        lw  = 2.4 if itype=='HBond' else 1.8

        # Nearest heavy atom
        ds  = np.sqrt((apx[heavy,0]-rx)**2+(apx[heavy,1]-ry)**2)
        ni  = heavy[np.argmin(ds)]
        lx, ly = apx[ni]

        dx_,dy_ = lx-rx, ly-ry
        d_      = math.sqrt(dx_**2+dy_**2)+1e-9
        ux_,uy_ = dx_/d_, dy_/d_
        ts      = math.sqrt((ow*ux_)**2+(oh*uy_)**2)
        x1_     = rx+ux_*ts
        y1_     = ry+uy_*ts
        x2_     = lx-ux_*C*0.014
        y2_     = ly-uy_*C*0.014

        ax.plot([x1_,x2_],[y1_,y2_],
                color=col, lw=lw, linestyle=ls,
                alpha=0.92, zorder=3,
                solid_capstyle='round')

        # Distance label
        if itype in ('HBond','SaltBridge','WaterBridge') and dist:
            mx_ = x1_+(x2_-x1_)*0.45
            my_ = y1_+(y2_-y1_)*0.45
            px_ = -uy_*C*0.028
            py_ =  ux_*C*0.028
            ax.text(mx_+px_, my_+py_,
                    f'{dist:.2f} Å',
                    fontsize=9, color=col,
                    fontweight='bold',
                    ha='center', va='center',
                    bbox=dict(facecolor='white',
                              edgecolor='none',
                              alpha=0.82, pad=1.2))

        # Residue oval
        ax.add_patch(Ellipse(
            (rx,ry), ow*2.2, oh*2.2,
            facecolor=col, edgecolor='white',
            linewidth=1.8, alpha=0.92, zorder=4))

        # Residue label
        parts  = rkey.strip().split()
        if len(parts)>=3:
            disp = f'{parts[0]}\n{parts[1]}:{parts[2]}'
        elif len(parts)==2:
            disp = f'{parts[0]}:{parts[1]}'
        else:
            disp = rkey
        ax.text(rx, ry, disp,
                ha='center', va='center',
                fontsize=int(C*0.0064),
                fontweight='bold',
                color='white', zorder=5,
                linespacing=1.2)

    # Title
    dg_str = f'   ΔG = {dg_val} kcal/mol' if dg_val else ''
    ax.set_title(
        f'{compound_name}{dg_str}\n'
        'AKT1 (PDB: 3O96)  ·  AutoDock Vina  ·  PLIP',
        fontsize=13, fontweight='bold',
        color='#1A3A5C', pad=12)

    # Legend
    seen = set(t for t,d in interactions.values())
    legs = []
    seen_l = set()
    for itype in ['HBond','Hydrophobic','PiStack',
                  'PiCation','SaltBridge','WaterBridge',
                  'HalogenBond','MetalComplex']:
        if itype not in seen:
            continue
        lbl = lbl_map.get(itype, itype)
        col = col_map.get(itype,'#778899')
        ls  = ls_map.get(itype,'--')
        if lbl not in seen_l:
            legs.append(Line2D([0],[0],color=col,lw=2.2,
                               linestyle=ls,label=lbl))
            legs.append(mpatches.Patch(
                facecolor=col,edgecolor='white',
                label=f'{lbl} residue'))
            seen_l.add(lbl)

    ax.legend(handles=legs, loc='lower left',
              fontsize=9.5, framealpha=0.92,
              edgecolor='#CCCCCC', fancybox=True)

    # Summary box
    counts = {}
    for t,d in interactions.values():
        counts[t] = counts.get(t,0)+1
    summary = '\n'.join(
        f'{lbl_map.get(t,t)}: {c}'
        for t,c in counts.items())
    summary += f'\nTotal: {len(interactions)}'
    ax.text(C*0.98, C*0.02, summary,
            ha='right', va='bottom',
            fontsize=9.5, color='#222222',
            bbox=dict(facecolor='white',
                      edgecolor='#CCCCCC',
                      alpha=0.92, pad=7,
                      boxstyle='round'))

    plt.tight_layout()
    safe = compound_name.replace(' ','_').replace('/','_')
    out  = 'LigPlot_' + safe + '.png'
    plt.savefig(out, dpi=300,
                bbox_inches='tight',
                facecolor='white')
    print('✅ Saved: ' + out)
    plt.show()

    # Cleanup
    try:
        os.remove(complex_pdb)
    except Exception:
        pass

    return out


# ── Test on one compound ──────────────────────────────────────────
TEST_SMILES = 'CCOC(=O)Oc1ccc2c(=O)c(-c3ccc(OC)cc3)coc2c1'

ligplot_plip(
    PROTEIN_PDB,
    TEST_PDB,
    ligand_smiles=TEST_SMILES,
    compound_name='Ethyl chloroformate',
    dg_val=-8.6)

# ================================================================
# DOWNLOAD ALL OUTPUTS — zip every generated PNG/CSV, organized
# ================================================================
import glob, os, zipfile
from datetime import datetime
from google.colab import files

# ── Patterns covering everything this pipeline can generate ───────
figure_patterns = [
    'LigPlot_*.png',                    # PLIP-based diagrams (Step 5)
    'interaction_*.png',                # ProLIF 2D interaction diagrams
    'Figure*.png',                      # Figure1, Figure3, Figure4, Figure5, Figure8...
]
table_patterns = [
    '*.csv',                            # docking rankings, interaction tables, etc.
]

# Collect unique matching files (avoid duplicates from overlapping globs)
figure_files = sorted(set(f for pat in figure_patterns for f in glob.glob(pat)))
table_files  = sorted(set(f for pat in table_patterns for f in glob.glob(pat)))

all_files = figure_files + table_files

if not all_files:
    print('⚠  No output files found matching expected patterns.')
    print('   Run the analysis/plotting steps first, then re-run this cell.')
else:
    print(f'📊 Figures found: {len(figure_files)}')
    for f in figure_files:
        size_kb = os.path.getsize(f) / 1024
        print(f'  • {f}  ({size_kb:.1f} KB)')

    print(f'\n📋 Tables found: {len(table_files)}')
    for f in table_files:
        size_kb = os.path.getsize(f) / 1024
        print(f'  • {f}  ({size_kb:.1f} KB)')

    # ── Build the zip with organized subfolders ─────────────────
    timestamp = datetime.now().strftime('%Y%m%d_%H%M%S')
    zip_name  = f'AKT1_formononetin_results_{timestamp}.zip'

    with zipfile.ZipFile(zip_name, 'w', zipfile.ZIP_DEFLATED) as zf:
        for f in figure_files:
            zf.write(f, arcname=f'figures/{os.path.basename(f)}')
        for f in table_files:
            zf.write(f, arcname=f'tables/{os.path.basename(f)}')

    zip_size_mb = os.path.getsize(zip_name) / (1024 * 1024)
    print(f'\n✅ Created: {zip_name}  ({zip_size_mb:.2f} MB, {len(all_files)} files total)')
    print(f'   Structure: figures/ ({len(figure_files)} files) + tables/ ({len(table_files)} files)')

    # ── Trigger browser download ────────────────────────────────
    files.download(zip_name)
    print('✅ Download triggered — check your browser downloads')
```
